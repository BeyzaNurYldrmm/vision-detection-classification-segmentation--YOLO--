```python
import cv2
import imutils
import numpy as np
from ultralytics import YOLO
```

```python
img=cv2.imread(r"dog.jpg")
```

```python
model = YOLO("yolov8n-cls.pt")
results = model(img, show=True)
```

```python
class_dict = results[0].names 
probs = results[0].probs.data.tolist()
```

```python
predicted_class_idx = np.argmax(probs)
predicted_class_name = class_dict[predicted_class_idx]
confidence = probs[predicted_class_idx]
```

```python
img = cv2.imread(r"C:\Users\Beyza Nur\Desktop\Projelerim\python\dog.jpg")
font=cv2.FONT_HERSHEY_SIMPLEX
font_scale=0.8
thickness=1
text = f"{predicted_class_name}, {confidence:.2f}"
cv2.putText(img, text, (15, 40), font, font_scale, (0, 0, 255), thickness)
cv2.imshow("Sınıflandırma Sonucu", img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
